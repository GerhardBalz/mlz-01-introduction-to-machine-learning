```python
import pandas as pd

df = pd.read_csv("car_fuel_efficiency_2026.csv")
df.shape
```

```python
# Q3: distinct fuel types
df["fuel_type"].nunique()
```

```python
# Q4: columns containing missing values
df.isna().any().sum()
```

```python
# Q5: maximum fuel efficiency for cars from Asia
df.loc[df["origin"] == "Asia", "fuel_efficiency_mpg"].max()
```

```python
# Q6: median horsepower before and after filling missing values
horsepower = df["horsepower"]
median_before = horsepower.median()
most_frequent = horsepower.mode().iloc[0]
median_after = horsepower.fillna(most_frequent).median()
print(f"Median before: {median_before}")
print(f"Most frequent: {most_frequent}")
print(f"Median after: {median_after}")
```

```python
# Q7: linear-regression weights from the first seven Asian cars
import numpy as np
X = df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]].head(7).to_numpy()
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
print("X shape:", X.shape)
print("w:", w)
print("sum:", w.sum())
```
